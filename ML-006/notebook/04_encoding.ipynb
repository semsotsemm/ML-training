```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
import numpy as np

file_path = "../data/processed/"
```

Для обучения модели был выбран алгоритм градиентного бустинга, значит значения не надо нормализовывать, а даты и категории надо преобразовать в числовые признаки, для преаброзования lithology используем OHE т.к. признаки не имеют иерархии, редкость категорий не несет физического смысла.

```python
train_df = pd.read_parquet(f"{file_path}train.parquet")

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

encoded_data = encoder.fit_transform(train_df[["lithology"]])

encoded_columns = encoder.get_feature_names_out(["lithology"])

train_df[encoded_columns] = encoded_data
train_df.drop(columns=["lithology"], inplace=True)
```

```python
train_df
```

Колонку well_id необходимо полностью удалить, во избежания переобучения

```python
train_df.drop(columns=["well_id"], inplace=True)
```

```python
train_df
```

И колонку со временем разобъем на числовые признаки: sin(t), cos(t), и прошедешее время, здесь важно использовать цикличную фукнцию, т.к. если использовать линейную алгоритм поймет порядок порядковых месяцев, но не поймет что между 12 и 1 месяц разница небольшая, чтобы избежать переобучения одной тригонометрической функцией добавляется сразу вторая.

```python
train_df["day_since_start"] = (train_df["t"] - train_df["t"].min()).dt.days

month = train_df["t"].dt.month

train_df["sin_t"] = np.sin(2 * np.pi * month / 12.0).round(4)
train_df["cos_t"] = np.cos(2 * np.pi * month / 12.0).round(4)

train_df.drop(columns=["t"], inplace=True)
```

```python
train_df.to_parquet(f"{file_path}encoded-train.parquet")
train_df.to_csv(f"{file_path}encoded-train.csv")
```
