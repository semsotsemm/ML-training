```python
import pandas as pd
import numpy as np

base_file_path = "../data/preprocessed/"
```

```python
df = pd.read_parquet(f"{base_file_path}merged_data.parquet")
```

```python
df
```

```python
date_value = 2 * np.pi * (df["concentration_t"].dt.month - 1) / 12

df["concentration_sin"] = np.sin(date_value)
df["concentration_cos"] = np.cos(date_value)
df["concentration_days_since_start"] = (df["concentration_t"] - pd.Timestamp("2020-01-01")).dt.days


date_value = 2 * np.pi * (df["head_date"].dt.month - 1) / 12

df["head_sin"] = np.sin(date_value)
df["head_cos"] = np.cos(date_value)
df["head_days_since_start"] = (df["head_date"] - pd.Timestamp("2020-01-01")).dt.days


date_value = 2 * np.pi * (df["recharge_date"].dt.month - 1) / 12

df["recharge_sin"] = np.sin(date_value)
df["recharge_cos"] = np.cos(date_value)
df["recharge_days_since_start"] = (df["recharge_date"] - pd.Timestamp("2020-01-01")).dt.days


date_value = 2 * np.pi * (df["well_commissioned"].dt.month - 1) / 12

df["well_commissioned_sin"] = np.sin(date_value)
df["well_commissioned_cos"] = np.cos(date_value)
df["well_commissioned_days_since_start"] = (df["well_commissioned"] - pd.Timestamp("2020-01-01")).dt.days


df = df.drop(columns=["concentration_t", "head_date", "recharge_date", "month", "well_commissioned", "well_id", "sample_id"])
```

```python
df.to_parquet(f"{base_file_path}finally_data.parquet")
```
