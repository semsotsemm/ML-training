Необходимо проверить координаты x,y,z в них очень много null а также неккоретных значений, удалить дубликаты, если у точки не задана хоть одна координата из нее нельзя извлечь никакой пользы так что такую точку будем просто удалять. Role аналогично проверяем на дубликаты, удаляем nan и некорретные значения. Аналогично поступаем и с zone. Commissioned приводим к единому формату даты pandas очищаем null. Дубликатами будем считать значения у которых совпадает x,y,z, commissioned.
