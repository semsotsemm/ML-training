```
import tifffile as tff
import matplotlib.pyplot as plt
import numpy as np

base_file_path = "../../data/raw/"
```

```
permeability = tff.imread(f"{base_file_path}permeability.tif")
```

```
permeability[permeability == -9999] = np.nan
```

```
plt.imshow(permeability, cmap="viridis")
plt.colorbar()

display(permeability.shape)
```

Permeability — карта проводимости верхнего слоя, более желтые участки — большая проводиомость , синии — меньшая. Шаг в карте равен 50, размер 60 на 80. Ячейка [0;0] имеет физические координаты (25;25).
