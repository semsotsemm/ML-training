```python
import pandas as pd
from preprocessing_utils import RAW_DIR, normalize_label, normalize_well_id, parse_date, parse_number, save_table

df = pd.read_csv(RAW_DIR / "well_surveys.csv")
df["well_id"] = normalize_well_id(df["well_id"])
df["survey_date"] = parse_date(df["survey_date"])
df["status"] = normalize_label(df["status"])
```

```python
for column in ["x", "y", "screen_top", "screen_bottom"]:
    df[column] = parse_number(df[column])

df = df.drop_duplicates()
```
