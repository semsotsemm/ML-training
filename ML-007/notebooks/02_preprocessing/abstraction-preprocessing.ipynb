```python
import pandas as pd
from preprocessing_utils import RAW_DIR, normalize_label, parse_date, parse_number, save_table

abstraction_df = pd.read_csv(RAW_DIR / "abstraction.csv")
abstraction_df["intake_name"] = normalize_label(abstraction_df["intake_name"])
abstraction_df["t"] = parse_date(abstraction_df["t"])

for column in ["x_center", "y_center", "total_abstraction_m3_day"]:
    abstraction_df[column] = parse_number(abstraction_df[column])

abstraction_df = abstraction_df.drop_duplicates()
```
