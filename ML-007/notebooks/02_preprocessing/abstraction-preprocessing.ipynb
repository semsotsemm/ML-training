```python
import pandas as pd


default_file_path = "../../data/raw/"
save_file_path = "../../data/iterim/"


def convert_integer_column(df_column: pd.Series):
    converted_column = df_column.astype("string").str.replace(" ", "", regex=False).str.replace(",", ".", regex=False).str.replace("\u00a0", "", regex=False).str.replace("—", "-", regex=False)
    converted_column = pd.to_numeric(converted_column, errors="coerce")
    return converted_column.round(2)


def convert_datetime_column(df_column: pd.Series):
    converted_column = pd.to_datetime(df_column, errors="coerce", format="mixed")
    return converted_column


def convert_string_column(df_column: pd.Series):
    convert_column = df_column.astype("string").str.lower().str.strip().str.replace(" ", "_")
    return convert_column
```

```python
abstraction_df = pd.read_csv(f"{default_file_path}abstraction.csv")
```

```python
display(abstraction_df)
display(abstraction_df.describe())
display(abstraction_df.info())
```

```python
abstraction_df["x_center"] = convert_integer_column(abstraction_df["x_center"])
abstraction_df["y_center"] = convert_integer_column(abstraction_df["y_center"])
abstraction_df["total_abstraction_m3_day"] = convert_integer_column(abstraction_df["total_abstraction_m3_day"])
abstraction_df["t"] = convert_datetime_column(abstraction_df["t"])
abstraction_df["intake_name"] = convert_string_column(abstraction_df["intake_name"])
```

```python
abstraction_df = abstraction_df.dropna(subset=["x_center", "y_center", "total_abstraction_m3_day", "t"])
```

```python
abstraction_df = abstraction_df.drop_duplicates(subset=["x_center", "y_center", "total_abstraction_m3_day", "t"])
```

```python
abstraction_df.to_parquet(f"{save_file_path}abstraction.parquet")
```
