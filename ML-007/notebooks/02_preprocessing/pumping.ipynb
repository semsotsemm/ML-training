```python
import pandas as pd
from preprocessing_utils import (
    RAW_DIR, latest_version, normalize_label, normalize_well_id,
    parse_date, parse_number, save_table,
)

df_pumping = pd.read_csv(RAW_DIR / "pumping.csv")
df_pumping["well_id"] = normalize_well_id(df_pumping["well_id"])
df_pumping["t"] = parse_date(df_pumping["t"])
df_pumping["units"] = normalize_label(df_pumping["units"])

df_pumping = latest_version(df_pumping, "record_id", "revision")
unit_scale = df_pumping["units"].map({"m3/day": 1, "l/s": 86.4})
df_pumping["pumping_rate"] = parse_number(df_pumping["pumping_rate"]) * unit_scale
df_pumping["units"] = "m3/day"
```
